**PYTHON DATA STRUCTURE**

**1. LIST**

```python
print("===== LIST DATASET =====")

marks = [70, 85, 90, 75, 85]

print("Original List:", marks)

# Access
print("First Value:", marks[0])
print("Last Value:", marks[-1])

# Insertion
marks.append(95)
print("After Append:", marks)

marks.insert(1, 80)
print("After Insert:", marks)

marks.extend([60, 65])
print("After Extend:", marks)

# Update
marks[0] = 100
print("After Update:", marks)

# Deletion
marks.remove(85)
print("After Remove:", marks)

marks.pop()
print("After Pop:", marks)

del marks[1]
print("After Delete:", marks)

# Search
print("Index of 90:", marks.index(90))

# Count
print("Count of 85:", marks.count(85))

# Sort
marks.sort()
print("Ascending:", marks)

marks.sort(reverse=True)
print("Descending:", marks)

# Reverse
marks.reverse()
print("Reversed:", marks)

# Length
print("Length:", len(marks))

# Maximum
print("Maximum:", max(marks))

# Minimum
print("Minimum:", min(marks))

# Sum
print("Sum:", sum(marks))

# Slicing
print("First Three Values:", marks[0:3])

# Membership
print("Is 90 Present:", 90 in marks)

# Copy
new_marks = marks.copy()

print("Copied List:", new_marks)

# Clear
new_marks.clear()

print("After Clear:", new_marks)
```

**3. SET**

```python
print("===== SET DATASET =====")

class_a = {"Ali", "Ahmed", "Sara", "Hamza"}
class_b = {"Sara", "Hamza", "Ayesha", "Umar"}

print("Class A:", class_a)
print("Class B:", class_b)

# Add
class_a.add("Bilal")

print("After Add:", class_a)

# Add Multiple Values
class_a.update(["Zain", "Usman"])

print("After Update:", class_a)

# Remove
class_a.remove("Bilal")

print("After Remove:", class_a)

# Discard
class_a.discard("Zain")

print("After Discard:", class_a)

# Union
union = class_a.union(class_b)

print("Union:", union)

# Intersection
intersection = class_a.intersection(class_b)

print("Intersection:", intersection)

# Difference
difference = class_a.difference(class_b)

print("Difference:", difference)

# Symmetric Difference
symmetric = class_a.symmetric_difference(class_b)

print("Symmetric Difference:", symmetric)

# Membership
print("Is Ali Present:", "Ali" in class_a)

# Length
print("Length:", len(class_a))

# Subset
set1 = {1, 2}
set2 = {1, 2, 3, 4}

print("Is Subset:", set1.issubset(set2))

# Superset
print("Is Superset:", set2.issuperset(set1))

# Disjoint
a = {1, 2}
b = {3, 4}

print("Are Disjoint:", a.isdisjoint(b))

# Pop
class_a.pop()

print("After Pop:", class_a)

# Copy
new_set = class_a.copy()

print("Copied Set:", new_set)

# Clear
new_set.clear()

print("After Clear:", new_set)
```

**4. DICTIONARY**

```python
print("===== DICTIONARY DATASET =====")

student = {
    "name": "Ali",
    "age": 20,
    "marks": 85,
    "result": "Pass"
}

print("Original Dictionary:", student)

# Access
print("Name:", student["name"])
print("Marks:", student["marks"])

# Get
print("Age:", student.get("age"))

# Insert New Data
student["city"] = "Abbottabad"

print("After Insertion:", student)

# Update Value
student["marks"] = 90

print("After Update:", student)

# Update Multiple Values
student.update({
    "age": 21,
    "result": "Excellent"
})

print("After Multiple Updates:", student)

# Keys
print("Keys:", student.keys())

# Values
print("Values:", student.values())

# Items
print("Items:", student.items())

# Membership
print("Is Name Key Present:", "name" in student)

# Length
print("Length:", len(student))

# Remove Using Pop
student.pop("city")

print("After Pop:", student)

# Remove Last Item
student.popitem()

print("After Pop Item:", student)

# Insert Default Value
student.setdefault("course", "AI")

print("After Set Default:", student)

# Copy
new_student = student.copy()

print("Copied Dictionary:", new_student)

# Delete
del student["age"]

print("After Delete:", student)

# Loop Through Keys
for key in student:
    print(key)

# Loop Through Values
for value in student.values():
    print(value)

# Loop Through Keys and Values
for key, value in student.items():
    print(key, ":", value)

# Clear
new_student.clear()

print("After Clear:", new_student)
```

**ALGORITHMS**

**1. Uniform Cost Search — UCS**

```python
import heapq
import networkx as nx
import matplotlib.pyplot as plt

graph = {
    "A": {"B": 4, "C": 2},
    "B": {"A": 4, "D": 5, "E": 12},
    "C": {"A": 2, "E": 10, "F": 4},
    "D": {"B": 5, "G": 6},
    "E": {"B": 12, "C": 10, "G": 3},
    "F": {"C": 4, "G": 5},
    "G": {"D": 6, "E": 3, "F": 5}
}


def ucs(graph, start, goal):
    priority_queue = [(0, start, [start])]
    visited = {}
    visited_order = []

    while priority_queue:
        cost, node, path = heapq.heappop(priority_queue)

        if node in visited and visited[node] <= cost:
            continue

        visited[node] = cost
        visited_order.append(node)

        if node == goal:
            return path, cost, visited_order

        for neighbor, edge_cost in graph[node].items():
            new_cost = cost + edge_cost
            new_path = path + [neighbor]

            heapq.heappush(
                priority_queue,
                (new_cost, neighbor, new_path)
            )

    return None, 0, visited_order


start = "A"
goal = "G"

path, total_cost, visited_order = ucs(graph, start, goal)

print("===== UNIFORM COST SEARCH =====")

print("Start Node:", start)
print("Goal Node:", goal)
print("Visited Order:", visited_order)
print("Optimal Path:", path)
print("Total Cost:", total_cost)


G = nx.Graph()

for node in graph:
    for neighbor, cost in graph[node].items():
        G.add_edge(node, neighbor, weight=cost)


pos = {
    "A": (0, 2),
    "B": (2, 3),
    "C": (2, 1),
    "D": (4, 3),
    "E": (4, 2),
    "F": (4, 0),
    "G": (6, 2)
}


path_edges = list(zip(path, path[1:]))

plt.figure(figsize=(9, 6))

nx.draw(
    G,
    pos,
    with_labels=True,
    node_size=2000
)

nx.draw_networkx_edge_labels(
    G,
    pos,
    edge_labels=nx.get_edge_attributes(G, "weight")
)

nx.draw_networkx_edges(
    G,
    pos,
    edgelist=path_edges,
    width=4
)

plt.title("Uniform Cost Search")
plt.show()
```

**2. A* SEARCH ALGORITHM**

```python
import heapq
import networkx as nx
import matplotlib.pyplot as plt

graph = {
    "A": {"B": 4, "C": 2},
    "B": {"A": 4, "D": 5, "E": 12},
    "C": {"A": 2, "E": 10, "F": 4},
    "D": {"B": 5, "G": 6},
    "E": {"B": 12, "C": 10, "G": 3},
    "F": {"C": 4, "G": 5},
    "G": {"D": 6, "E": 3, "F": 5}
}


heuristic = {
    "A": 9,
    "B": 8,
    "C": 7,
    "D": 6,
    "E": 3,
    "F": 5,
    "G": 0
}


def a_star(graph, heuristic, start, goal):
    priority_queue = [(heuristic[start], 0, start, [start])]
    best_cost = {start: 0}
    visited_order = []

    while priority_queue:
        f_cost, g_cost, node, path = heapq.heappop(priority_queue)

        if g_cost > best_cost.get(node, float("inf")):
            continue

        visited_order.append(node)

        if node == goal:
            return path, g_cost, visited_order

        for neighbor, edge_cost in graph[node].items():
            new_g = g_cost + edge_cost
            new_f = new_g + heuristic[neighbor]

            if new_g < best_cost.get(neighbor, float("inf")):
                best_cost[neighbor] = new_g

                heapq.heappush(
                    priority_queue,
                    (
                        new_f,
                        new_g,
                        neighbor,
                        path + [neighbor]
                    )
                )

    return None, 0, visited_order


start = "A"
goal = "G"

path, total_cost, visited_order = a_star(
    graph,
    heuristic,
    start,
    goal
)


print("===== A STAR SEARCH =====")

print("Start Node:", start)
print("Goal Node:", goal)
print("Heuristic Dataset:", heuristic)
print("Visited Order:", visited_order)
print("Optimal Path:", path)
print("Total Cost:", total_cost)


G = nx.Graph()

for node in graph:
    for neighbor, cost in graph[node].items():
        G.add_edge(node, neighbor, weight=cost)


pos = {
    "A": (0, 2),
    "B": (2, 3),
    "C": (2, 1),
    "D": (4, 3),
    "E": (4, 2),
    "F": (4, 0),
    "G": (6, 2)
}


node_labels = {}

for node in G.nodes():
    node_labels[node] = (
        node +
        "\nh=" +
        str(heuristic[node])
    )


path_edges = list(zip(path, path[1:]))

plt.figure(figsize=(9, 6))

nx.draw(
    G,
    pos,
    labels=node_labels,
    with_labels=True,
    node_size=2200
)

nx.draw_networkx_edge_labels(
    G,
    pos,
    edge_labels=nx.get_edge_attributes(G, "weight")
)

nx.draw_networkx_edges(
    G,
    pos,
    edgelist=path_edges,
    width=4
)

plt.title("A Star Search Algorithm")
plt.show()
```
